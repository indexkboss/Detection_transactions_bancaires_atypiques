```python
# 10. Grouping all features into a dictionary for better organization
feature_groups = {
    "Main transaction features": transaction_main_features,
    "Card features": card_features,
    "Address features": addr_features,
    "Distance features": dist_features,
    "Email features": email_features,
    "C features": c_features,
    "D features": d_features,
    "M features": m_features,
    "V features": v_features
}

for group_name, features in feature_groups.items():
    print(f"{group_name}: {len(features)} features")
```

```python
#Question importante : toutes les colonnes sont-elles dans ces groupes ?
grouped_features = []

for features in feature_groups.values():
    grouped_features.extend(features)

ungrouped_features = [
    column
    for column in train_transaction_df.columns
    if column not in grouped_features
]

print("Ungrouped features:", ungrouped_features)
```

```python
#Réponse : toutes les colonnes sont dans ces groupes, il n'y a pas de colonnes non regroupées.
```

```python
# 11. Missing values analysis by feature group
group_missing_summary = {}

for group_name, features in feature_groups.items():

    features_existing = [
        column
        for column in features
        if column in train_transaction_df.columns
    ]

    if len(features_existing) > 0:

        average_missing = (
            train_transaction_df[features_existing]
            .isnull()
            .mean()
            .mean()
            * 100
        )

        group_missing_summary[group_name] = average_missing

        
group_missing_summary = pd.Series(
    group_missing_summary
).sort_values(ascending=False)

group_missing_summary
```

```python
very_high_missing_transaction = (
    missing_transaction_percentage[
        missing_transaction_percentage > 90
    ]
    .sort_values(ascending=False)
)

print(
    "Number of columns with more than 90% missing values:",
    len(very_high_missing_transaction)
)

very_high_missing_transaction
```

```python
# 3. Comparer les montants selon isFraud:
#objectif:

# Pour chaque groupe (isFraud = 0 et isFraud = 1) :
# Médiane → où se situe le montant typique
# Dispersion → est-ce que les montants sont concentrés ou étalés (via la boîte et les moustaches)
# Valeurs extrêmes → jusqu'où ça va

# Le boxplot est parfait pour ça : il résume une distribution en 5 nombres sans dessiner chaque point.

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))

sns.boxplot(
    data=train_transaction_df,
    x="isFraud",
    y="TransactionAmt"
)

plt.title("Distribution de TransactionAmt selon isFraud")
plt.xlabel("isFraud (0 = non-fraude, 1 = fraude)")
plt.ylabel("Montant de la transaction")
plt.show()
```

```python
#le boxplot est écrasé donc la médiane est très basse solution:
plt.figure(figsize=(8, 6))
sns.boxplot(data=train_transaction_df, x="isFraud", y="TransactionAmt")
plt.yscale("log")
plt.title("TransactionAmt selon isFraud (échelle log)")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(
    data=train_transaction_df,
    x="isFraud",
    y="TransactionAmt",
    showfliers=False # cache les outliers pour comparer médiane + dispersion entre les deux groupes
)
plt.title("TransactionAmt selon isFraud (sans outliers)")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# gauche : avec outliers, échelle log
sns.boxplot(data=train_transaction_df, x="isFraud", y="TransactionAmt", ax=axes[0])
axes[0].set_yscale("log")
axes[0].set_title("Avec outliers (échelle log)")

# droite : sans outliers, échelle normale
sns.boxplot(data=train_transaction_df, x="isFraud", y="TransactionAmt",
            showfliers=False, ax=axes[1])
axes[1].set_title("Sans outliers")

plt.tight_layout()
plt.show()
```

Le boxplot compare la distribution des montants entre transactions frauduleuses et non frauduleuses. Comme les données brutes sont écrasées par les valeurs extrêmes, deux vues complémentaires ont été produites : une échelle logarithmique (pour voir les outliers) et une vue sans outliers (pour comparer le corps des distributions).

Médiane : la médiane des transactions frauduleuses (~75) est supérieure à celle des non-fraudes (~68.5), ce qui indique que les fraudes portent en général sur des montants plus élevés.

Dispersion : la boîte des fraudes est plus large (35–161) que celle des non-fraudes (44–120), ce qui signifie que les montants frauduleux sont plus variés.

Valeurs extrêmes : les deux groupes contiennent des outliers, mais les non-fraudes atteignent des montants beaucoup plus élevés (jusqu'à 30 000), tandis que les fraudes plafonnent autour de 5 000 dans cette vue.

Conclusion : le montant de la transaction est un signal modérément discriminant — les fraudes ont tendance à être légèrement plus élevées et plus dispersées, mais il y a un fort chevauchement entre les deux distributions, ce qui signifie que le montant seul ne suffit pas à détecter la fraude.

```python
# 3 Créer une information temporelle cyclique :
'''
Si nous créons :
RelativeHour = 23
RelativeHour = 0
numériquement, le modèle peut considérer 23 et 0 comme très éloignés

Alors qu'en réalité 23h → 00h sont très proches

Pour une approche ML, nous pouvons donc représenter l'heure avec : cos et sin
'''

seconds_per_day = 24 * 60 * 60

df["RelativeHour"] = (
    (
        df["TransactionDT"]
        % seconds_per_day
    )
    // 3600
)

df["RelativeHour"].value_counts().sort_index()
```

```python
# Solution : Créer HourSin et HourCos
df["HourSin"] = np.sin(
    2 * np.pi * df["RelativeHour"] / 24
)

df["HourCos"] = np.cos(
    2 * np.pi * df["RelativeHour"] / 24
)
```

```python
# 4 Transformer TransactionAmt :
'''Pourquoi log1p :Parce qu'elle calcule log(1 + x) et évite les problèmes liés à log(0)'''

df["LogTransactionAmt"] = np.log1p(
    df["TransactionAmt"]
)
```

Pour les écarts bruts énormes, après transformation logarithmique, les valeurs deviennent beaucoup plus compactes.
Cela peut aider certains modèles à mieux gérer une distribution fortement asymétrique.

⚠️ Avec XGBoost, cette transformation n'est pas obligatoirement nécessaire. Nous la créons surtout parce qu'elle peut être utile et nous pourrons comparer son intérêt.

```python
#Numerical: imputation numérique (remplacer les valeurs numériques manquantes)

#Méthodes simples (statistiques) :
    # La moyenne : 
        #Remplace la valeur vide par la moyenne de la colonne.Utile si les données sont bien réparties.
    # La médiane : 
        #Remplace par la valeur du milieu. Préférable à la moyenne s'il y a des valeurs extrêmes (outliers).
    # Une constante : 
        #Remplace par un chiffre fixe (souvent 0 ou -1)     

#Méthodes avancées :
    #KNN (K plus proches voisins) : 
        # Estime la valeur manquante en se basant sur les observations les plus proches géométriquement 
        # ou mathématiquement dans l'espace des variables.
    # Modèles prédictifs (Régression) : 
        # Utilise les autres colonnes du dataset comme variables explicatives pour prédire la valeur 
        # numérique absente.
```

```python
#Categorical: imputation catégorielle (encoding)

#Méthodes simples :
    # Le mode (la valeur la plus fréquente) : 
            # C'est la méthode statistique par défaut pour le catégoriel.
    # Création d'une catégorie "Inconnu" ou "Manquant" : 
            # C'est très utile si l'absence d'information a une signification en soi 
            # par exemple un client qui n'a pas renseigné son numéro de téléphone fixe

# Méthodes avancées :
    # KNN (K plus proches voisins) : 
            # Cherche les lignes de données les plus similaires et attribue la catégorie majoritaire chez eux 
    # Classification prédictive : 
            # Utilise un algorithme (comme un arbre de décision ou une régression logistique) 
            # pour prédire la catégorie manquante en fonction des autres caractéristiques de la ligne.
```

```python
#PIPELINE NUMERIQUE:
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)
'''
justification choix mediane:
La médiane est généralement plus robuste aux valeurs extrêmes
tandis que la moyenne sera influencer par outliers
'''
```

```python
#PIPELINE CATEGORIEL:
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),  
        #OneHotEncoder: converts categorical text data into a binary list so a computer can process it
        #la variable est nominale → il n'y a pas d'ordre entre les catégories ; 
        # et le nombre de catégories n'est pas énorme
        ("encoder", OneHotEncoder(
                #handle_unknown c'est pour ne pas crée pas de nouvelle colonne
                handle_unknown="ignore",
                sparse_output=True
            )
        )
    ]
)
```

```python
#combined:
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features)
    ]
)
```

```python
'''
                 X_train_baseline
                    │
          ┌─────────┴─────────┐
          ↓                   ↓
    numeric_features    categorical_features
          ↓                   ↓
    numeric_pipeline:   categorical_pipeline:
          ↓                   ↓
       médiane             mode imputation
                              ↓
                         One-Hot Encoding
          └─────────┬─────────┘
                    ↓
              X_processed
'''
```

```python
#FIT uniquement sur Train
X_train_processed = preprocessor.fit_transform(
    X_train_baseline
)
'''
avec 'fit':

    Le pipeline apprend à partir du train :
    les médianes 
    la catégorie la plus fréquente 
    les catégories existantes

transform :
    applique ces règles
'''
```

```python
# Transformer Validation et Test
X_validation_processed = preprocessor.transform(
    X_validation_baseline
)

X_test_processed = preprocessor.transform(
    X_test_baseline
)
print("X_train_processed shape:", X_train_processed.shape)
print("X_validation_processed shape:", X_validation_processed.shape)
print("X_test_processed shape:", X_test_processed.shape)
# on fait pas fit_transform(X_test) ou  fit_transform(X_validation) 
# Sinon on introdui des informations du futur dans le preprocessing
```

## 13. XGBoost:

```python
'''on va passer à un modèle beaucoup plus puissant

La régression logistique recherche essentiellement une relation relativement simple entre les variables et le résultat.
Mais la fraude peut dépendre d'interactions complexes 
(ex: montant élevé + catégorie particulière + type de carte + ...)

XGBoost, basé sur des arbres de décision boosting, peut capturer le genre de relations non linéaires
'''
```

```python
#1. installer et verifier XGBoost:
import xgboost
print(xgboost.__version__)
```

```python
#2. "scale_pos_weight" :
'''
on a un déséquilibre
XGBoost possède un paramètre scale_pos_weight qui donnera davantage de poids aux exemples frauduleux

La formule généralement est :
scale_pos_weight = nombre_classe_0 / nombre_classe_1
'''

n_negative = (y_train == 0).sum()
n_positive = (y_train == 1).sum()

scale_pos_weight = n_negative / n_positive

print(
    "scale_pos_weight:",
    scale_pos_weight
)
```

```python
#3. construire model:
# scale_pos_weight= 5
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=250,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="aucpr",
    scale_pos_weight=scale_pos_weight,
    tree_method="hist",
    random_state=42,
    n_jobs=-1
)
```

## n_estimators=250
Nombre d'arbres construits
## max_depth=6
Contrôle la profondeur maximale des arbres.
Une profondeur trop grande peut favoriser l'overfitting
## learning_rate=0.05
Contrôle la contribution de chaque nouvel arbre.
Un taux plus faible signifie généralement qu'il faut davantage d'arbres
## subsample=0.8
Chaque arbre utilise environ 80 % des observations disponibles pour sa construction
Cela peut contribuer à limiter l'overfitting
## colsample_bytree=0.8
Chaque arbre utilise environ 80 % des variables
## eval_metric="aucpr"
Nous demandons à XGBoost de surveiller AUC-PR plutôt qu'une métrique basée uniquement sur l'accuracy
## tree_method="hist"
Cela demande une méthode d'entraînement basée sur des histogrammes
C'est utile avec un dataset de cette taille

## 16. Construction du Risk Score

```python
# objectif :
# probabilité XGBoost + score Isolation Forest -> Risk Score -> Low / Medium / High
''' 
                    Transaction
                         │
              ┌──────────┴──────────┐
              ↓                     ↓
          XGBoost             Isolation Forest
              ↓                     ↓
     FraudProbability        AnomalyScore
              │                     │
              └──────────┬──────────┘
                         ↓
                    Risk Score
                         ↓
                niveau de risque
'''
```

```python
#1. objectif: 
# normaliser isolation forest score entre 0 et 1 pour le comparer à la probabilité de fraude de XGBoost:
'''xgb_validation_proba est déjà entre 0 et 1'''
'''isolation forest nous donne '''
```

```python
#2. Transformer le score d'anomalie:
from sklearn.preprocessing import MinMaxScaler

anomaly_scaler = MinMaxScaler().fit(
    (-isolation_forest.decision_function(X_train_processed)).reshape(-1, 1)
)
validation_anomaly_normalized = np.clip(
    anomaly_scaler.transform((-if_validation_score).reshape(-1, 1)).ravel(), 0, 1
)
#Isolation Forest : score faible = plus anormal 
# on veut le contraire : score élevé = plus anormal donc : -if_validation_score

#fit: trouve le min et max du score d'anomalie
#transform: applique la formule de normalisation min-max
#ravel: transforme le tableau 2D en 1D
```

```python
#verification:
print("Min :", validation_anomaly_normalized.min())
print("Max :", validation_anomaly_normalized.max())
print("Mean:", validation_anomaly_normalized.mean())
```

```python
#1 = transaction la plus anormale de ton ensemble de validation 
#0 = la moins anormale
```

```python
#3. construire un premier Risk Score:
'''baseline, prenons : 
70 % XGBoost
30 % Isolation Forest'''

risk_score = (
    0.70 * xgb_validation_proba
    + 0.30 * validation_anomaly_normalized
)
risk_score_100 = risk_score * 100

print("Min :", risk_score_100.min())
print("Max :", risk_score_100.max())
```

```python
#4. créer un DataFrame pour analyser les résultats:
validation_risk_results = validation_df[
    [
        "TransactionID",
        "TransactionAmt",
        "ProductCD",
        "isFraud"
    ]
].copy()

validation_risk_results["FraudProbability"] = (
    xgb_validation_proba
)

validation_risk_results["AnomalyScore"] = (
    if_validation_score
)

validation_risk_results["AnomalyNormalized"] = (
    validation_anomaly_normalized
)

validation_risk_results["RiskScore"] = (
    risk_score_100
)
```

```python
# 1.10 Transformer l'explication SHAP en langage métier
feature_descriptions = {
    "TransactionAmt": "Montant de la transaction",
    "LogTransactionAmt": "Montant transformé logarithmiquement",
    "ProductCD": "Catégorie de transaction",
    "card4": "Réseau de carte bancaire",
    "card6": "Type de carte bancaire",
    "addr1": "Information d'adresse principale",
    "addr2": "Information d'adresse secondaire",
    "dist1": "Distance associée à la transaction",
    "dist2": "Distance secondaire",
    "P_emaildomain": "Domaine email de l'acheteur",
    "R_emaildomain": "Domaine email du destinataire",
    "RelativeHour": "Heure relative de la transaction",
    "HourSin": "Composante temporelle sinusoïdale",
    "HourCos": "Composante temporelle cosinus",
    "HasIdentity": "Disponibilité des informations d'identité",
    "DeviceType": "Type d'appareil utilisé"
}

def describe_shap_feature(feature_name):
    original_name = feature_name.split("__")[-1]

    if original_name in feature_descriptions:
        return feature_descriptions[original_name]

    # pour les var One-Hot Encoding
    for feature, description in feature_descriptions.items():
        if original_name.startswith(feature + "_"):
            category = original_name[len(feature) + 1:]
            return f"{description} - catégorie {category}"

    return original_name


business_explanation = transaction_explanation.copy()

business_explanation["BusinessDescription"] = (
    business_explanation["Feature"]
    .apply(describe_shap_feature)
)

business_explanation["Impact"] = np.where(
    business_explanation["SHAPValue"] > 0,
    "Augmente la sortie du modèle en faveur de la fraude",
    np.where(
        business_explanation["SHAPValue"] < 0,
        "Diminue la sortie du modèle en faveur de la fraude",
        "Contribution nulle"
    )
)

business_explanation = business_explanation.sort_values(
    "AbsSHAP",
    ascending=False
)

display(
    business_explanation[
        [
            "Feature",
            "BusinessDescription",
            "FeatureValue",
            "SHAPValue",
            "Impact"
        ]
    ].head(10)
)
```

une valeur SHAP positive augmente la sortie expliquée par le modèle, et une valeur négative la diminue. Selon la configuration de TreeExplainer, cette sortie peut être un score brut ou des log-odds, pas directement une variation en pourcentage de probabilité. Ce tableau explique donc la contribution au modèle, pas une cause certaine de fraude.

```python
#2.1 Explication descriptive d'une anomalie avec Isolation Forest

# Sélectionner la transaction ayant le score d'anomalie le plus élevé
# (decision_function le plus faible = observation plus anormale)

anomaly_position = np.argmin(if_validation_score)

anomaly_transaction = validation_df.iloc[anomaly_position]

print("Transaction sélectionnée :")
print("TransactionID :", anomaly_transaction["TransactionID"])
print("Montant :", anomaly_transaction["TransactionAmt"])
print("Score Isolation Forest :", if_validation_score[anomaly_position])


# Comparer les caractéristiques avec le jeu d'entraînement

anomaly_explanation = []

for feature in baseline_features:

    value = anomaly_transaction[feature]

    train_values = train_df[feature]

    if pd.api.types.is_numeric_dtype(train_values):

        median_value = train_values.median()

        q1 = train_values.quantile(0.25)
        q3 = train_values.quantile(0.75)

        iqr = q3 - q1

        if pd.isna(value):
            unusualness = np.nan
            description = "Valeur manquante"

        else:
            if iqr == 0:
                unusualness = float(value != median_value)
            else:
                unusualness = min(
                    abs(value - median_value) / (iqr + 1e-8),
                    1
                )

            description = (
                f"Valeur observée : {value} | "
                f"Médiane train : {median_value:.3f}"
            )

    else:

        category_frequency = (
            train_values
            .value_counts(normalize=True, dropna=False)
        )

        frequency = category_frequency.get(value, 0)

        unusualness = 1 - frequency

        description = (
            f"Catégorie observée : {value} | "
            f"Fréquence train : {frequency:.4f}"
        )

    anomaly_explanation.append({
        "Feature": feature,
        "ObservedValue": value,
        "Description": description,
        "UnusualnessScore": unusualness
    })


anomaly_explanation_df = pd.DataFrame(anomaly_explanation)

anomaly_explanation_df = anomaly_explanation_df.sort_values(
    "UnusualnessScore",
    ascending=False
)

display(anomaly_explanation_df.head(10))
```

## 18. Optimisation du seuil de décision

```python
''' 
Jusqu'ici, XGBoost utilise le seuil par défaut de 0.5 pour convertir ses probabilités en prédictions 0/1. 
Pour un problème déséquilibré, il est utile d'étudier un seuil sur le jeu de validation.

On va choisir un seuil qui maximise le F1-score sur Validation, puis le conserver pour Test. 
Le jeu Test ne servira pas à choisir le seuil.
'''
```

```python
#1. Calculer les seuils possibles
from sklearn.metrics import precision_recall_curve

precision_values, recall_values, thresholds = precision_recall_curve(
    y_validation,
    xgb_validation_proba
)

# Les seuils ont une longueur de 1 inférieure
# aux tableaux precision et recall

f1_values = (
    2 * precision_values[:-1] * recall_values[:-1]
    /
    (
        precision_values[:-1] + recall_values[:-1] + 1e-10
    )
)

best_threshold_index = np.argmax(f1_values)

best_threshold = thresholds[best_threshold_index]

print("Seuil sélectionné :", best_threshold)
print("Precision :", precision_values[best_threshold_index])
print("Recall :", recall_values[best_threshold_index])
print("F1-score :", f1_values[best_threshold_index])
```

```python
#2. Visualiser le F1 selon le seuil 
plt.figure(figsize=(10, 5))

plt.plot(
    thresholds,
    f1_values,
    label="F1-score"
)

plt.axvline(
    best_threshold,
    color="red",
    linestyle="--",
    label=f"Seuil sélectionné = {best_threshold:.3f}"
)

plt.xlabel("Seuil de décision")
plt.ylabel("F1-score")
plt.title("Optimisation du seuil XGBoost sur Validation")
plt.legend()
plt.grid()
plt.show()
```

```python
#3. Appliquer le seuil choisi sur Validation
xgb_validation_pred_tuned = (
    xgb_validation_proba >= best_threshold
).astype(int)

print("Classification avec le seuil sélectionné :")

print(
    classification_report(
        y_validation,
        xgb_validation_pred_tuned,
        zero_division=0
    )
)

print("Matrice de confusion :")

print(
    confusion_matrix(
        y_validation,
        xgb_validation_pred_tuned
    )
)
```
